# Мощность, MDE и размер выборки
Меняйте значения в блоке **ВВОД ДАННЫХ**, затем выполните **Run All**.
Графики строятся при `SHOW_PLOTS = True`.


```python
from power_mde import *
import numpy as np
import matplotlib.pyplot as plt
```

## ВВОД ДАННЫХ


```python
N = 30               # количество испытаний
P0 = 0.50            # вероятность успеха при H0
P_ALT = 0.60         # вероятность успеха при H1
ALPHA = 0.05         # уровень значимости
TARGET_POWER = 0.80  # желаемая мощность
MDE_TARGET = 0.10    # ожидаемый эффект в долях (+10 п.п.)
SHOW_PLOTS = True    # строить графики
```

## Вычисления
Мощность — вероятность обнаружить эффект при выбранных параметрах. MDE — минимальный обнаруживаемый эффект при заданной мощности. Размер выборки — минимальное число испытаний для указанного эффекта.


```python
print(f'Мощность: {binomial_power(N, P0, P_ALT, ALPHA):.2%}')
print(f'MDE: {binomial_mde(N, P0, ALPHA, TARGET_POWER):.4f} (абс. доля)')
print(f'Размер выборки: {required_sample_size(P0, MDE_TARGET, ALPHA, TARGET_POWER)}')
```

## Графики


```python
if SHOW_PLOTS:
    sizes = np.arange(10, 600, 10)
    powers = [binomial_power(int(n), P0, P_ALT, ALPHA) for n in sizes]
    plt.figure(figsize=(9, 5))
    plt.plot(sizes, powers, label='Мощность')
    plt.axhline(TARGET_POWER, ls='--', label='Целевая мощность')
    plt.xlabel('Размер выборки')
    plt.ylabel('Мощность')
    plt.ylim(0, 1)
    plt.legend()
    plt.show()

    effects = np.linspace(0, max(0, 1-P0), 120)
    powers_by_effect = [binomial_power(N, P0, P0 + float(d), ALPHA) for d in effects]
    plt.figure(figsize=(9, 5))
    plt.plot(effects * 100, powers_by_effect, label='Мощность')
    plt.axhline(TARGET_POWER, ls='--', label='Целевая мощность')
    plt.axvline(100*binomial_mde(N, P0, ALPHA, TARGET_POWER), ls=':', label='MDE')
    plt.xlabel('Эффект, процентные пункты')
    plt.ylabel('Мощность')
    plt.ylim(0, 1)
    plt.legend()
    plt.show()
```
