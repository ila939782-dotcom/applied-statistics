# statistical test: comparison of the critical region and the p-value

Правосторонний биномиальный критерий: критическая область, p-value и итоговое решение.


```python
from scipy.stats import binom
import numpy as numpy
from matplotlib import pyplot


# параметры задачи
n = 30
mu = 0.5
q = 19
alpha = 0.05


# биномиальное распределение при H0
binom_h0 = binom(
    n=n,
    p=mu
)


# функция критической области
def make_binom_criterion(n, mu=0.5, alpha=0.05):
    '''Строит критерий для задачи с доставкой

    Параметры:
        n: количество доставок в эксперименте
        mu: вероятность успеха в нулевой гипотезе
        alpha: уровень значимости критерия

    Возвращает:
        C для критерия S = {Q >= C}
    '''
    binom_h0 = binom(n=n, p=mu)

    # функция квантиля считается через ppf (percent point function)
    # квантиль точка между обычной и критической областью
    q_crit = binom_h0.ppf(1 - alpha)

    return int(q_crit + 1)


# функция p-value
def pvalue_binom(n, mu, q):
    '''Считает pvalue для задачи с доставкой

    Параметры:
        n: количество доставок в эксперименте
        mu: вероятность успеха при H0
        q: количество успешных доставок

    Возвращает:
        pvalue для правостороннего критерия
    '''
    binom_h0 = binom(n=n, p=mu)

    # cdf (кумулятивная функция распределения)
    # считаем вероятность получить q или больше
    return 1 - binom_h0.cdf(q - 1)


# считаем критическое значение
critical_value = make_binom_criterion(
    n=n,
    mu=mu,
    alpha=alpha
)


# считаем p-value
p_value = pvalue_binom(
    n=n,
    mu=mu,
    q=q
)


# координата
x_grid = numpy.arange(0, n + 1)

# pmf (probability mass function)
# высота столбцов на графике
probs = binom_h0.pmf(x_grid)


pyplot.figure(figsize=(16, 8))


# обычная область
normal_reg = x_grid < critical_value

pyplot.vlines(
    x_grid[normal_reg],
    0,
    probs[normal_reg],
    linewidth=12.0,
    color='black',
    label='Обычная область'
)


# отдельно изобразим критическую область критерия
crit_reg = x_grid >= critical_value

pyplot.vlines(
    x_grid[crit_reg],
    0,
    probs[crit_reg],
    linewidth=12.0,
    color='red',
    label=f'Критическая область: Q >= {critical_value}'
)


# наблюдаемое значение
pyplot.axvline(
    q,
    color='blue',
    linestyle='--',
    linewidth=3,
    label=f'Наблюдаемое значение Q = {q}'
)


# подписываем p-value
pyplot.text(
    q + 0.5,
    max(probs) * 0.8,
    f'p-value = {p_value:.4f}',
    fontsize=14
)


pyplot.title('Binomial distribution', fontsize=20)
pyplot.xlabel('Количество успехов', fontsize=14)
pyplot.ylabel('Вероятность', fontsize=14)
pyplot.legend(fontsize=14)
pyplot.show()


# конечный результат
print('----- РЕЗУЛЬТАТ -----')
print(f'p-value = {p_value:.6f}')
print(f'alpha = {alpha}')
print(f'Критическое значение = {critical_value}')
print(f'Наблюдаемое значение Q = {q}')

if p_value <= alpha:
    print('p-value <= alpha')
    print('Решение: reject H0')
else:
    print('p-value > alpha')
    print('Решение: do not reject H0')
```
